欧几里得算法

```python
# 1. 实现 gcd 函数
def gcd(m, n):
    # 欧几里得算法：不断用较小数除较大数，直到余数为0
    while n != 0:
        m, n = n, m % n
    return abs(m) # 返回绝对值，防止负数输入导致结果为负

# 2. 评估数字对
print("--- 测试 GCD ---")
print(f"gcd(2018, 2019) = {gcd(2018, 2019)}")
print(f"gcd(1200, 300) = {gcd(1200, 300)}")
print(f"gcd(5040, 60) = {gcd(5040, 60)}")
```

问题： 如果一个或两个参数是负数，你的函数会做什么？这种行为有意义吗？
回答： 如果传入负数，由于 Python 中 % 运算符（取模）的结果符号由除数决定，m % n 的结果总是非负的。因此，即使在循环中遇到负数，n 也会逐渐变为 0，最终 m 会停留在两个数绝对值的最大公约数上（或者其相反数）。为了严谨，我在函数末尾加了 abs(m) 确保返回正数。这种行为是有意义的，因为最大公约数定义在正整数集上。

近似欧拉数 e

```python
import math

# 1. euler_limit
def euler_limit(n):
    return (1 + 1/n) ** n

# 2. euler_infinite_sum
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1 / factorial
    return total

# 3. euler_approx
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    factorial = 1
    k = 0
    while abs(total - true_e) > epsilon:
        if k > 0:
            factorial *= k
        total += 1 / factorial
        k += 1
    return total

# 4. print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"k={k}: {euler_infinite_sum(k)}")

print("--- 测试 euler 相关函数 ---")
print(f"euler_limit(1000) = {euler_limit(1000)}")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}")
print(f"euler_approx(0.01) = {euler_approx(0.01)} (真实值约为 {math.exp(1)})")
print("euler_infinite_sum 表 (n=5):")
print_euler_sum_table(5)
```

测试整数属性

```python
# 1. is_power_of_2 (不使用 math.sqrt)
def is_power_of_2(n):
    if n <= 0:
        return False
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

# 2. is_power (推广版)
def is_power(b, n):
    if b <= 1 or n <= 0:
        return False
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

print("--- 测试整数属性 ---")
print(f"is_power_of_2(8): {is_power_of_2(8)}")
print(f"is_power_of_2(10): {is_power_of_2(10)}")
print(f"is_power(3, 27): {is_power(3, 27)}")
print(f"is_power(3, 28): {is_power(3, 28)}")
```
