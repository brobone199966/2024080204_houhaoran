字符串的乐趣

```python
# 1. is_palindrome
def is_palindrome(s):
    # 忽略空格并转为小写
    s_clean = s.replace(" ", "").lower()
    return s_clean == s_clean[::-1]

# 2. is_abecedarian
def is_abecedarian(s):
    # 忽略空格并转为小写
    s_clean = s.replace(" ", "").lower()
    # 遍历检查相邻字符是否按字母顺序排列
    for i in range(len(s_clean) - 1):
        if s_clean[i] > s_clean[i+1]:
            return False
    return True

# 3. remove_vowels
def remove_vowels(s):
    vowels = "aeiouAEIOU"
    # 列表推导式保留非元音字符
    return "".join([char for char in s if char not in vowels])

# 测试代码
print("--- 字符串测试 ---")
print(f"is_palindrome('taco cat'): {is_palindrome('taco cat')}")
print(f"is_palindrome('hello'): {is_palindrome('hello')}")
print(f"is_abecedarian('abcd efgh xyz'): {is_abecedarian('abcd efgh xyz')}")
print(f"is_abecedarian('dog'): {is_abecedarian('dog')}")
print(f"remove_vowels('audio'): {remove_vowels('audio')}")
```

列表的乐趣

```python
# 1. list_reverse
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是一个列表")
    return lst[::-1]

# 2. is_sorted
def is_sorted(seq):
    # 只遍历一次，比较相邻元素
    for i in range(len(seq) - 1):
        if seq[i] > seq[i+1]:
            return False
    return True

# 测试代码
print("\n--- 列表测试 ---")
print(f"list_reverse([1, 2, 3]): {list_reverse([1, 2, 3])}")
try:
    list_reverse("not a list")
except TypeError as e:
    print(f"捕获异常: {e}")
print(f"is_sorted([1, 2, 2, 5]): {is_sorted([1, 2, 2, 5])}")
print(f"is_sorted([1, 3, 2]): {is_sorted([1, 3, 2])}")
```

作为向量的元组

```python
# 辅助函数：检查元组内的元素是否全为数字
def _is_num_tuple(t):
    return all(isinstance(x, (int, float)) for x in t)

# 1. vec_scalar_mult
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple) or not _is_num_tuple(t):
        raise TypeError("第一个参数必须是数字元组")
    if not isinstance(s, (int, float)):
        raise TypeError("第二个参数必须是数字")
    return tuple(x * s for x in t)

# 2. vec_inner_product
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个元组的长度必须一致")
    return float(sum(x * y for x, y in zip(t1, t2)))

# 3. check_valid_mx
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if not _is_num_tuple(row):
            return False
    # 检查所有内层元组长度是否一致
    if len(mx) > 0:
        first_len = len(mx[0])
        for row in mx:
            if len(row) != first_len:
                return False
    return True

# 4. mx_vec_mult
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise ValueError("输入的不是有效的矩阵")
    if not isinstance(v, tuple) or not _is_num_tuple(v):
        raise TypeError("向量必须是数字元组")
    if len(mx[0]) != len(v):
        raise ValueError("矩阵的列数必须等于向量的维度")
    
    result = []
    for row in mx:
        # 矩阵的每一行与向量做内积
        result.append(vec_inner_product(row, v))
    return tuple(result)

# 测试代码
print("\n--- 向量与矩阵测试 ---")
print(f"vec_scalar_mult((1, 2), 3): {vec_scalar_mult((1, 2), 3)}")
print(f"vec_inner_product((1, 2), (3, 4)): {vec_inner_product((1, 2), (3, 4))}")
mx = ((1, 2), (3, 4))
v = (5, 6)
print(f"check_valid_mx({mx}): {check_valid_mx(mx)}")
print(f"mx_vec_mult({mx}, {v}): {mx_vec_mult(mx, v)}")
```

更多向量的乐趣（面向对象）

```python
class Vector:
    # 1. 初始化函数
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise TypeError("维度必须是非负整数")
        
        self.dim = dim
        
        if elements is None:
            # 默认行为：全0向量
            self.elements = tuple([0] * dim)
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("元素必须是列表或元组")
            if len(elements) != dim:
                raise ValueError(f"元素个数 ({len(elements)}) 必须与维度 ({dim}) 一致")
            if not _is_num_tuple(elements):
                raise TypeError("元素必须全是数字")
            self.elements = tuple(elements)

    # 3. 比较运算符 (基于元组的字典序比较)
    def _check_comparison(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与 Vector 对象比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    def __eq__(self, other):
        self._check_comparison(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_comparison(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_comparison(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_comparison(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_comparison(other)
        return self.elements >= other.elements

    # 4. 点积
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("维度必须一致")
        return float(sum(x * y for x, y in zip(self.elements, other.elements)))

    # 5. 乘法运算符
    def __mul__(self, other):
        # 标量乘法 (v * 2)
        if isinstance(other, (int, float)):
            return Vector(self.dim, [x * other for x in self.elements])
        # 向量逐元素乘法 (v * w)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度必须一致")
            return Vector(self.dim, [x * y for x, y in zip(self.elements, other.elements)])
        else:
            return NotImplemented

    def __rmul__(self, other):
        # 标量右乘 (2 * v)
        if isinstance(other, (int, float)):
            return self.__mul__(other)
        return NotImplemented

    # 方便打印查看
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

# 测试代码
print("\n--- Vector 类测试 ---")
v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))
v3 = Vector(2) # 默认(0, 0)

print(f"v1: {v1}")
print(f"v1 < v2: {v1 < v2}")
print(f"v1 == Vector(2, [2, 3]): {v1 == Vector(2, [2, 3])}")
print(f"v1.dot(v2): {v1.dot(v2)}")
print(f"v1 * 2: {v1 * 2}")
print(f"2 * v1: {2 * v1}")
print(f"v1 * v2: {v1 * v2}")
```

维度和元素应是类属性还是实例属性？说明选择的理由。
回答：维度和元素应该是实例属性。
理由：类属性是所有实例共享的。而在本题中，每一个 Vector 对象（例如二维向量和三维向量）都有自己独特的维度和坐标值。如果定义为类属性，修改一个向量的数据会影响其他所有向量，这违背了面向对象封装和独立性的原则。
